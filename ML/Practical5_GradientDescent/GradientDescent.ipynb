# ML Practical 5 — Gradient Descent to Find Local Minima

**Objective:** Implement the Gradient Descent algorithm to find the local minima of a function.  
**Example Function:** y = (x + 3)²  
**Starting Point:** x = 2  
**Known Minima:** x = -3 (where y = 0)


---
## Step 1 — Import Libraries

We only need `numpy` for math and `matplotlib` for visualization.  
No dataset is needed — gradient descent works purely on the mathematical function.

```python
import numpy as np
import matplotlib.pyplot as plt

print('Libraries imported')
```

---
## Step 2 — Define the Function and Its Derivative

**Function:** y = (x + 3)²  
This is a simple parabola with a global minimum at x = -3.

**Derivative (Gradient):** dy/dx = 2(x + 3)  
The derivative tells us the slope of the function at any point x.  
Gradient descent uses this slope to decide which direction to move x to reduce y.

```python
# The function: y = (x + 3)^2
def f(x):
    return (x + 3) ** 2

# Derivative of the function: dy/dx = 2*(x + 3)
def df(x):
    return 2 * (x + 3)

# Verify the minimum is at x = -3
print(f'f(-3) = {f(-3)}   <- should be 0 (the minimum)')
print(f'df(-3) = {df(-3)}  <- gradient should be 0 at minimum')
```

---
## Step 3 — Gradient Descent Algorithm

**How it works:**
1. Start at an initial x value.
2. Compute the gradient (slope) at current x.
3. Move x in the **opposite direction** of the gradient (downhill).
4. Repeat until the change in x is negligibly small (convergence).

**Update rule:** `x_new = x_old - learning_rate × gradient`

**Parameters:**
- `learning_rate` (α): controls how big each step is. Too large → overshoots. Too small → very slow.
- `tolerance`: stops when the change in x is smaller than this value.

```python
def gradient_descent(start_x, learning_rate, max_iterations, tolerance=1e-6):
    """
    Perform gradient descent to find the minimum of f(x).
    Returns the history of x and y values at each step.
    """
    x          = start_x
    x_history  = [x]
    y_history  = [f(x)]

    for i in range(max_iterations):
        gradient = df(x)                    # slope at current x
        x_new    = x - learning_rate * gradient   # step in opposite direction

        x_history.append(x_new)
        y_history.append(f(x_new))

        # Stop if the step size is smaller than tolerance (converged)
        if abs(x_new - x) < tolerance:
            print(f'Converged at iteration {i+1}')
            break

        x = x_new

    return x, f(x), x_history, y_history


# Parameters
start_x       = 2       # Starting point as given in the problem
learning_rate = 0.1     # Step size
max_iterations = 1000

x_min, y_min, x_hist, y_hist = gradient_descent(start_x, learning_rate, max_iterations)

print(f'\nStarting x     : {start_x}')
print(f'Minimum x found: {x_min:.6f}')
print(f'Minimum y value: {y_min:.6f}')
print(f'Steps taken    : {len(x_hist) - 1}')
```

---
## Step 4 — Visualize the Descent Path on the Curve

We plot the function y = (x+3)² as a smooth curve and overlay the actual steps taken by gradient descent.  
- Each red dot is the position of x at one iteration.
- You can see the dots starting at x=2 and progressively moving toward x=-3.
- The green star marks the final minimum found.

```python
x_curve = np.linspace(-8, 4, 300)
y_curve = f(x_curve)

plt.figure(figsize=(10, 6))
plt.plot(x_curve, y_curve, 'b-', linewidth=2, label='y = (x+3)²')

# Plot the gradient descent steps
plt.scatter(x_hist, y_hist, color='red', s=50, zorder=5, label='GD Steps')
plt.plot(x_hist, y_hist, 'r--', alpha=0.4)

# Mark start and minimum
plt.scatter(x_hist[0],  y_hist[0],  color='orange', s=150, zorder=6, label=f'Start (x={start_x})')
plt.scatter(x_min,      y_min,      color='green',  s=200, marker='*', zorder=7, label=f'Minimum (x≈{x_min:.4f})')

plt.title('Gradient Descent on y = (x + 3)²')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

---
## Step 5 — Plot y Value Over Iterations

This chart shows how the value of y decreases with each iteration.  
A steep initial drop followed by flattening confirms gradient descent is working correctly — large steps early on, tiny steps near the minimum.

```python
plt.figure(figsize=(10, 4))
plt.plot(range(len(y_hist)), y_hist, color='steelblue', linewidth=2)
plt.title('y Value vs Iteration (Convergence Curve)')
plt.xlabel('Iteration')
plt.ylabel('y = f(x)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

---
## Step 6 — Effect of Learning Rate

The learning rate has a big impact on gradient descent behavior:
- **Too small** (e.g., 0.01) → very slow convergence, many iterations needed
- **Just right** (e.g., 0.1) → smooth, efficient convergence
- **Too large** (e.g., 0.9) → overshoots, may oscillate or diverge

We compare how fast each learning rate converges.

```python
learning_rates = [0.01, 0.1, 0.5, 0.9]
colors         = ['green', 'steelblue', 'orange', 'red']

plt.figure(figsize=(12, 5))

for lr, color in zip(learning_rates, colors):
    _, _, _, y_h = gradient_descent(start_x=2, learning_rate=lr, max_iterations=100)
    plt.plot(range(len(y_h)), y_h, color=color, linewidth=2, label=f'lr = {lr}')

plt.title('Effect of Learning Rate on Convergence')
plt.xlabel('Iteration')
plt.ylabel('y = f(x)')
plt.ylim(-1, 30)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

---
## Step 7 — Print Step-by-Step Trace

We print the first 10 iterations to see exactly how x moves step by step toward the minimum.

```python
print(f'{"Iteration":<12} {"x":<15} {"y = f(x)":<15} {"Gradient"}')
print('-' * 55)

x_trace = start_x
for i in range(10):
    grad  = df(x_trace)
    y_val = f(x_trace)
    print(f'{i:<12} {x_trace:<15.6f} {y_val:<15.6f} {grad:.6f}')
    x_trace = x_trace - 0.1 * grad
```

**Key Concepts:**
- **Gradient** = slope of the function at a point. Positive slope means we move left (decrease x). Negative slope means we move right (increase x).
- **Learning Rate** = how big each step is. The most critical hyperparameter.
- **Convergence** = when the steps become so small the algorithm stops (change < tolerance).
- **Update Rule:** x = x − α × (dy/dx)
